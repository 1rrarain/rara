```python

# SOAL 1 - OPERASI ARITMATIKA

A = 25
B = 4

print("===== OPERASI ARITMATIKA =====")
print("Nilai A =", A)
print("Nilai B =", B)

# Penjumlahan
print("A + B =", A + B)

# Pengurangan
print("A - B =", A - B)

# Perkalian
print("A * B =", A * B)

# Pembagian
print("A / B =", A / B)

# Pembagian bulat
print("A // B =", A // B)

# Modulus (sisa hasil bagi)
print("A % B =", A % B)

# Pangkat
print("A ** B =", A ** B)
```

```python

# SOAL 2 - LUAS SEGITIGA

print("===== MENGHITUNG LUAS SEGITIGA =====")

# Memasukkan nilai alas dan tinggi
alas = float(input("Masukkan alas segitiga: "))
tinggi = float(input("Masukkan tinggi segitiga: "))

# Menghitung luas segitiga
luas = 0.5 * alas * tinggi

# Menampilkan hasil
print("------------------------------")
print("Alas segitiga   :", alas)
print("Tinggi segitiga :", tinggi)
print("Luas segitiga   :", luas)
```

```python

# SOAL 3 - GAJI KARYAWAN

# Menentukan komponen gaji
gaji_pokok = 3500000
tunjangan = 750000
potongan = 250000

# Menghitung gaji bersih
gaji_bersih = gaji_pokok + tunjangan - potongan

# Menampilkan rincian gaji
print("===== SLIP GAJI KARYAWAN =====")
print("Gaji Pokok  : Rp", format(gaji_pokok, ",").replace(",", "."))
print("Tunjangan   : Rp", format(tunjangan, ",").replace(",", "."))
print("Potongan    : Rp", format(potongan, ",").replace(",", "."))
print("------------------------------")
print("Gaji Bersih : Rp", format(gaji_bersih, ",").replace(",", "."))
print("==============================")
```

```python

# SOAL 4 - SISA PEMBAGIAN

print("=== PROGRAM SISA PEMBAGIAN ===")

# Meminta pengguna memasukkan bilangan
bilangan = int(input("Masukkan sebuah bilangan: "))

# Menghitung hasil pembagian dengan 7
hasil_bagi = bilangan // 7

# Menghitung sisa pembagian menggunakan operator %
sisa = bilangan % 7

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN ===")
print("Bilangan       :", bilangan)
print("Pembagi        : 7")
print("Hasil bagi     :", hasil_bagi)
print("Sisa pembagian :", sisa)
```

```python

# SOAL 5 - ANALISIS EKSPRESI

print("=== PROGRAM ANALISIS EKSPRESI ===")

# Menentukan nilai perhitungan
angka1 = 15
angka2 = 5
angka3 = 3
angka4 = 10
angka5 = 2

# Menghitung ekspresi sesuai prioritas operator
hasil = (angka1 + angka2) * angka3 - angka4 / angka5

# Menampilkan proses dan hasil
print("Ekspresi: (15 + 5) * 3 - 10 / 2")
print("--------------------------------")

print("Langkah 1: 15 + 5 =", angka1 + angka2)
print("Langkah 2: 20 * 3 =", (angka1 + angka2) * angka3)
print("Langkah 3: 10 / 2 =", angka4 / angka5)

print("--------------------------------")
print("Hasil akhir =", hasil)
```
